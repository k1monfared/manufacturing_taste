# Cultural Market Simulation - Exploration

This notebook explores the basic behavior of the cultural market simulation.

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
import matplotlib.pyplot as plt
from cultural_market import CulturalMarket, DEFAULT_PARAMS
from cultural_market.visualization import plot_quality_success_scatter
```

## Basic Simulation Run

```python
# Create and run a simulation
market = CulturalMarket({
    'n_producers': 100,
    'n_consumers': 500,
    't_canon': 50
})
market.initialize(seed=42)
market.run()

# Show metrics
metrics = market.compute_metrics()
print("Simulation Metrics:")
for key, value in metrics.items():
    print(f"  {key}: {value:.4f}" if isinstance(value, float) else f"  {key}: {value}")
```

## Quality vs Success Scatterplot

```python
qualities = np.array([p.quality for p in market.producers])
successes = np.array([p.cumulative_success for p in market.producers])
canonical = np.array([p.canonical for p in market.producers])

fig, ax = plt.subplots(figsize=(10, 6))
plot_quality_success_scatter(qualities, successes, canonical, ax=ax)
plt.show()
```

## Comparing Independent vs Social Conditions

```python
# Independent condition (no social influence)
market_ind = CulturalMarket({
    'n_producers': 100,
    'n_consumers': 500,
    't_canon': 50,
    'si_gamma': 0.0  # No social influence
})
market_ind.initialize(seed=42)
market_ind.run()

# Social condition
market_soc = CulturalMarket({
    'n_producers': 100,
    'n_consumers': 500,
    't_canon': 50,
    'si_gamma': 0.5  # With social influence
})
market_soc.initialize(seed=42)
market_soc.run()

print("Independent Condition:")
metrics_ind = market_ind.compute_metrics()
print(f"  Quality-Success Correlation: {metrics_ind['quality_success_correlation']:.4f}")
print(f"  Gini Coefficient: {metrics_ind['gini_coefficient']:.4f}")

print("\nSocial Condition:")
metrics_soc = market_soc.compute_metrics()
print(f"  Quality-Success Correlation: {metrics_soc['quality_success_correlation']:.4f}")
print(f"  Gini Coefficient: {metrics_soc['gini_coefficient']:.4f}")
```

## Dynamics Over Time

```python
# Plot how metrics evolve over time
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gini over time
times = [snapshot.t for snapshot in market.history]
ginis = [snapshot.gini for snapshot in market.history]
axes[0].plot(times, ginis)
axes[0].set_xlabel('Time')
axes[0].set_ylabel('Gini Coefficient')
axes[0].set_title('Inequality Over Time')

# Quality-Success Correlation over time
corrs = [snapshot.quality_success_corr for snapshot in market.history]
axes[1].plot(times, corrs)
axes[1].set_xlabel('Time')
axes[1].set_ylabel('Quality-Success Correlation')
axes[1].set_title('Meritocracy Over Time')

plt.tight_layout()
plt.show()
```
