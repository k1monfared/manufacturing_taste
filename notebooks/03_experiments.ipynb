# Running Experiments

This notebook runs and visualizes the five main experiments.

```python
import sys
sys.path.insert(0, '../src')

from cultural_market.experiments import (
    experiment_salganik_replication,
    experiment_counterfactual,
    experiment_variance_decomposition,
)
from cultural_market.visualization import (
    plot_condition_comparison,
    plot_variance_decomposition,
    plot_canonical_by_decile,
)
import matplotlib.pyplot as plt
```

## Experiment 1: Salganik Replication

```python
# Run with smaller n_runs for quick testing
result = experiment_salganik_replication(n_runs=20, seed=42)

print("Salganik Replication Results:")
print(f"\nIndependent condition:")
print(f"  Correlation: {result.metrics['independent']['quality_success_corr_mean']:.3f}")
print(f"  Gini: {result.metrics['independent']['gini_mean']:.3f}")

print(f"\nSocial condition:")
print(f"  Correlation: {result.metrics['social']['quality_success_corr_mean']:.3f}")
print(f"  Gini: {result.metrics['social']['gini_mean']:.3f}")

print(f"\nComparison:")
print(f"  Gini ratio: {result.metrics['comparison']['gini_ratio']:.3f}")
print(f"  Correlation difference: {result.metrics['comparison']['corr_difference']:.3f}")
```

```python
fig = plot_condition_comparison(result)
plt.show()
```

## Experiment 2: Counterfactual Canon Formation

```python
result_cf = experiment_counterfactual(n_runs=50, quality_seed=42)

print("Counterfactual Results:")
print(f"  Counterfactual distance: {result_cf.metrics['counterfactual_distance']:.3f}")
print(f"  Canonical probability variance: {result_cf.metrics['canonical_probability_variance']:.4f}")
print(f"  Mean canonical prob: {result_cf.metrics['mean_canonical_prob']:.3f}")
print(f"  Max canonical prob: {result_cf.metrics['max_canonical_prob']:.3f}")
```

```python
fig = plot_canonical_by_decile(result_cf)
plt.show()
```

## Experiment 3: Variance Decomposition

```python
result_vd = experiment_variance_decomposition(n_runs=20, seed=42)

print("Variance Decomposition:")
for source, value in result_vd.metrics['decomposition'].items():
    print(f"  {source}: {value*100:.1f}%")
```

```python
fig = plot_variance_decomposition(result_vd)
plt.show()
```

## Summary

The experiments demonstrate:

1. **Salganik Replication**: Social influence reduces quality-success correlation and increases inequality
2. **Counterfactual**: Canonical status is path-dependent - different capital allocations lead to different canons
3. **Variance Decomposition**: Success is driven by a combination of quality, capital, and social influence
