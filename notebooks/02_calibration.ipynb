# Calibration Against Salganik Data

This notebook demonstrates calibrating the simulation against empirical data from the MusicLab experiments.

```python
import sys
sys.path.insert(0, '../src')

from cultural_market.calibration import (
    load_calibration_targets,
    simulate_independent_condition,
    simulate_social_condition,
    calibrate
)
from cultural_market.market import DEFAULT_PARAMS
```

## Load Calibration Targets

```python
targets = load_calibration_targets()

print("Calibration Targets from Salganik et al. (2006):")
for target in targets:
    print(f"  {target.name}:")
    print(f"    Target value: {target.value}")
    print(f"    Acceptable range: [{target.range_low}, {target.range_high}]")
```

## Test Current Parameters

```python
# Test with default parameters
print("Testing default parameters...")

ind_results = simulate_independent_condition(DEFAULT_PARAMS, n_runs=5)
print(f"\nIndependent condition:")
print(f"  Quality-Success Correlation: {ind_results['quality_success_correlation']:.4f}")
print(f"  Gini Coefficient: {ind_results['gini_coefficient']:.4f}")

soc_results = simulate_social_condition(DEFAULT_PARAMS, n_runs=5)
print(f"\nSocial condition:")
print(f"  Quality-Success Correlation: {soc_results['quality_success_correlation']:.4f}")
print(f"  Gini Coefficient: {soc_results['gini_coefficient']:.4f}")

print(f"\nGini ratio (social/independent): {soc_results['gini_coefficient']/ind_results['gini_coefficient']:.4f}")
```

## Run Calibration

Note: Full calibration can take a while. For quick testing, reduce `n_runs`.

```python
# Run calibration (use small n_runs for quick testing)
print("Running calibration (this may take a few minutes)...")
result = calibrate(n_runs=3, seed=42)

print(f"\nCalibration complete!")
print(f"Final loss: {result['final_loss']:.4f}")
print(f"\nOptimized parameters:")
for name, value in result['optimized_values'].items():
    print(f"  {name}: {value:.4f}")
```

## Verify Calibrated Parameters

```python
# Test calibrated parameters
calibrated_params = result['calibrated_params']

ind_results = simulate_independent_condition(calibrated_params, n_runs=10)
soc_results = simulate_social_condition(calibrated_params, n_runs=10)

print("Results with calibrated parameters:")
print(f"\nIndependent condition:")
print(f"  Quality-Success Correlation: {ind_results['quality_success_correlation']:.4f} (target: 0.65)")

print(f"\nSocial condition:")
print(f"  Quality-Success Correlation: {soc_results['quality_success_correlation']:.4f} (target: 0.40)")

gini_ratio = soc_results['gini_coefficient'] / ind_results['gini_coefficient']
print(f"\nGini ratio: {gini_ratio:.4f} (target: 1.30)")
```
